# Task 25: Basic Data Cleaning Workflow
**Track:** AI & ML Track | Level 1  
**Dataset:** Titanic Dataset (`titanic.csv`)  
**Objective:** Perform fundamental data cleaning including missing-value handling, duplicate removal, and data-type correction.

```python
# Import required libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## Step 1: Load and Inspect the Original Dataset
First, we load the raw dataset and inspect its structure, data types, column names, and initial rows to understand the data layout.

```python
# Load the original dataset
df = pd.read_csv('titanic.csv')

# Display the first 5 rows
print("--- First 5 Rows of Original Dataset ---")
display(df.head())

# Display dataset shape and basic info
print("\n--- Dataset Info ---")
df.info()
```

## Step 2: Check for Missing Values and Duplicates
Next, we systematically check for missing values across all columns and identify any duplicate rows in the dataset.

```python
# Check for missing values count per column
print("--- Missing Values Count ---")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

# Check for duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")
```

## Step 3: Handle Missing Values
Based on our inspection:
- **`Age`**: Contains missing values. We will fill them using the median age.
- **`Cabin`**: Contains a very high percentage of missing values. We can either drop it or create a binary indicator, but for basic cleaning, dropping or handling sparse columns is standard. Let's drop `Cabin` due to excessive missing data.
- **`Embarked`**: Contains a couple of missing values. We will fill them with the most frequent value (mode).
- **`Fare`**: Check and fill if any missing.

```python
# Create a copy for cleaning
df_cleaned = df.copy()

# 1. Handle 'Age' missing values using the median
median_age = df_cleaned['age'].median()
df_cleaned['age'] = df_cleaned['age'].fillna(median_age)

# 2. Handle 'Embarked' missing values using the mode
mode_embarked = df_cleaned['embarked'].mode()[0]
df_cleaned['embarked'] = df_cleaned['embarked'].fillna(mode_embarked)

# 3. Drop 'Cabin' column due to excessive missing values (>75% missing)
if 'cabin' in df_cleaned.columns:
    df_cleaned = df_cleaned.drop(columns=['cabin'])

# 4. Drop 'PassengerId', 'Name', and 'Ticket' if they aren't needed for modeling, 
# or keep them if tracking identifiers. Let's keep them for record integrity unless redundant, 
# but verify remaining missing values.
print("--- Missing Values After Handling ---")
print(df_cleaned.isnull().sum())
```

## Step 4: Remove Duplicates and Correct Data Types
We check again for any duplicate entries and ensure that categorical and numerical columns have appropriate data types (e.g., converting object types to categories or numeric types).

```python
# Remove duplicate rows if any exist
initial_rows = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates()
final_rows = len(df_cleaned)

print(f"Duplicates removed: {initial_rows - final_rows}")

# Correct data types
# Convert categorical features to 'category' type
categorical_cols = ['Sex', 'Embarked', 'Pclass', 'Survived']
for col in categorical_cols:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].astype('category')

print("\n--- Data Types After Correction ---")
print(df_cleaned.dtypes)
```

## Step 5: Verify and Save the Cleaned Dataset
We perform a final verification check on our cleaned dataframe and export it as `titanic_cleaned.csv`.

```python
# Verify cleaned dataset summary
print("--- Cleaned Dataset Summary ---")
display(df_cleaned.describe(include='all'))

# Save the cleaned dataset to a new CSV file
df_cleaned.to_csv('titanic_cleaned.csv', index=False)
print("\nSuccess! Cleaned dataset saved as 'titanic_cleaned.csv'.")
```

## Interview Questions & Answers

### Q1: What is data cleaning?
**Answer:** Data cleaning is the process of detecting, correcting, or removing corrupt, inaccurate, duplicate, improperly formatted, or incomplete records from a dataset. It ensures data quality, consistency, and reliability before it is used for analysis or machine learning modeling.

### Q2: Why is data cleaning important before machine learning?
**Answer:** 
- **Prevents Model Bias & Errors:** Algorithms learn patterns from data; garbage in results in garbage out (GIGO). Missing values or corrupt entries can cause runtime errors or bias model weights.
- **Improves Accuracy:** Clean data helps algorithms converge faster and yield higher predictive accuracy.
- **Feature Integrity:** Ensures features are in the correct data format (numerical, categorical) required by machine learning frameworks.

### Q3: What problems can poor-quality data cause?
**Answer:**
- **Model Failure:** Missing or malformed values can cause pipeline crashes or exception errors during training.
- **Overfitting/Underfitting:** Outliers and noise can lead models to learn spurious correlations rather than true underlying patterns.
- **Misleading Insights:** Flawed aggregations or duplicate records skew statistical metrics (mean, median, standard deviation), leading to incorrect business or analytical conclusions.
